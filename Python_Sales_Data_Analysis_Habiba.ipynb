# 📊 Sales Data Analysis Using Python

## Project Overview

This project analyzes sales data using Python, pandas, and Matplotlib.
The analysis focuses on sales performance, product quantities, and key business insights.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone", "Tablet"],
    "Quantity": [2, 5, 3, 1, 4, 2],
    "Price": [800, 500, 300, 800, 500, 300]
}

df = pd.DataFrame(data)

df
```

```python
df["Sales"] = df["Quantity"] * df["Price"]

df
```

```python
df["Sales"].sum()
```

```python
df.groupby("Product")["Sales"].sum()
```

```python
sales_by_product = df.groupby("Product")["Sales"].sum()

sales_by_product.plot(kind="bar")

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

```python
df.groupby("Product")["Quantity"].sum()
```

```python
quantity_by_product = df.groupby("Product")["Quantity"].sum()

quantity_by_product.plot(kind="bar")

plt.title("Quantity Sold by Product")
plt.xlabel("Product")
plt.ylabel("Quantity")

plt.show()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.sort_values("Sales", ascending=False)
```

```python
df.groupby("Product")["Sales"].sum().sort_values(ascending=False)
```

```python
sales_by_product = df.groupby("Product")["Sales"].sum()

sales_percentage = (sales_by_product / sales_by_product.sum()) * 100

sales_percentage
```

```python
quantity_by_product = df.groupby("Product")["Quantity"].sum()

quantity_by_product.sort_values(ascending=False)
```
